3. Data Transformation

Convert the rating and approximate cost columns to numeric types, drop unnecessary text-heavy columns, and add useful analytical columns.


```python
# Convert the rating text such as "4.1/5" into a numeric rating.
df["rating"] = pd.to_numeric(
    df["rate"].astype("string").str.replace("/5", "", regex=False).str.strip(),
    errors="coerce",
)

# Convert approximate cost text into a numeric value.
df["approx_cost"] = pd.to_numeric(
    df["approx_cost(for two people)"].astype("string").str.replace(",", "", regex=False),
    errors="coerce",
).astype(float)
df["votes"] = pd.to_numeric(df["votes"], errors="coerce").astype(float)
print("Rating, cost, and votes converted to numeric columns.")
df[["rating", "approx_cost", "votes"]].head()
```

```python
# Drop columns that are not required for this analysis.
columns_to_drop = ["url", "address", "phone", "reviews_list", "menu_item", "rate", "approx_cost(for two people)"]
df = df.drop(columns=columns_to_drop)
print("Dropped unnecessary columns:")
print(columns_to_drop)
print("Remaining columns:")
print(df.columns.tolist())
```

```python
# Add derived columns for analysis.
df["online_order_flag"] = df["online_order"].eq("Yes").astype(int)
df["table_booking_flag"] = df["book_table"].eq("Yes").astype(int)
df["engagement_score"] = df["votes"] * df["rating"]
print("New columns added successfully.")
df[["online_order_flag", "table_booking_flag", "engagement_score"]].head()
```

4. Handling Null Values

Identify null values and fill numeric columns using mean imputation wherever applicable. Categorical columns use mode imputation because a numeric mean is not meaningful for text.


```python
print("Checking null values before filling")
df.isnull().sum().sort_values(ascending=False)
```

```python
# Mean imputation for numeric columns.
numeric_columns = ["rating", "approx_cost", "votes", "engagement_score"]
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

# Mode imputation for categorical columns.
categorical_columns = [
    "name", "online_order", "book_table", "location", "rest_type",
    "dish_liked", "cuisines", "listed_in(type)", "listed_in(city)",
]
for column in categorical_columns:
    mode = df[column].mode(dropna=True)
    if not mode.empty:
        df[column] = df[column].fillna(mode.iloc[0])

print("Null values filled using mean/mode imputation.")
```

```python
print("Checking null values after filling")
df.isnull().sum().sort_values(ascending=False)
```

```python
# Add a second transformed measure based on engagement.
df["double_engagement"] = df["engagement_score"].apply(lambda value: value * 2)
df[["engagement_score", "double_engagement"]].head()
```

5. Insight Generation

Generate insights using grouping, aggregation, counting, sorting, and comparison of restaurant attributes.


```python
# Group locations and find their average rating.
location_rating = df.groupby("location")["rating"].mean().sort_values(ascending=False)
print("Average rating by location:")
print(location_rating.head(10))
```

```python
# Group restaurant types and find their average rating.
type_rating = df.groupby("rest_type")["rating"].mean().sort_values(ascending=False)
print("Average rating by restaurant type:")
print(type_rating.head(10))
```

```python
# Group restaurant types and find their average cost.
type_cost = df.groupby("rest_type")["approx_cost"].mean().sort_values(ascending=False)
print("Average cost by restaurant type:")
print(type_cost.head(10))
```

```python
# Compare restaurants that offer online ordering.
online_order_analysis = df.groupby("online_order")[["rating", "votes", "approx_cost"]].mean()
print("Online-order comparison:")
print(online_order_analysis)
```

```python
# Count restaurants in each location and sort in descending order.
location_counts = df.groupby("location")["name"].count().sort_values(ascending=False)
print("Top locations by restaurant count:")
print(location_counts.head(10))
```

```python
# Count restaurants in each restaurant type and sort in descending order.
type_counts = df.groupby("rest_type")["name"].count().sort_values(ascending=False)
print("Top restaurant types:")
print(type_counts.head(10))
```

```python
# Compare restaurants with and without table booking.
booking_analysis = df.groupby("book_table")[["rating", "votes", "approx_cost"]].mean()
print("Table-booking comparison:")
print(booking_analysis)
```

```python
# Display the most popular restaurants by number of votes.
most_voted = df.sort_values("votes", ascending=False)
print(most_voted[["name", "location", "rating", "votes"]].head(10))
```

```python
# Display the highest-rated restaurants after imputation.
highest_rated = df.sort_values(["rating", "votes"], ascending=[False, False])
print(highest_rated[["name", "location", "rating", "votes"]].head(10))
```

6. Data Visualization

Visualize restaurant counts, ratings, costs, and relationships using appropriate charts.


```python
# Bar graph for the top 10 locations by restaurant count.
top_locations = location_counts.head(10).sort_values()
top_locations.plot(kind="barh", figsize=(9, 5))
plt.xlabel("Number of Restaurants")
plt.ylabel("Location")
plt.title("Top 10 Locations by Restaurant Count")
plt.tight_layout()
plt.show()
```

```python
# Bar graph for the top 10 restaurant types.
top_types = type_counts.head(10).sort_values()
top_types.plot(kind="barh", figsize=(9, 5))
plt.xlabel("Number of Restaurants")
plt.ylabel("Restaurant Type")
plt.title("Top 10 Restaurant Types")
plt.tight_layout()
plt.show()
```

```python
# Bar graph comparing online-order availability.
online_counts = df["online_order"].value_counts()
online_counts.plot(kind="bar", figsize=(7, 5))
plt.xlabel("Online Order")
plt.ylabel("Number of Restaurants")
plt.title("Restaurants by Online-Order Availability")
plt.xticks(rotation=0, ha="center")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot for rating versus votes.
plt.figure(figsize=(8, 5))
plt.scatter(df["rating"], df["votes"], alpha=0.25, s=12)
plt.xlabel("Rating")
plt.ylabel("Votes")
plt.title("Restaurant Rating versus Votes")
plt.tight_layout()
plt.show()
```

```python
# Histogram for the distribution of restaurant ratings.
plt.figure(figsize=(8, 5))
plt.hist(df["rating"], bins=10, color="#f96b06", edgecolor="#1f77b4")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurant Ratings")
plt.tight_layout()
plt.show()
```

```python
# Histogram for the approximate cost for two people.
plt.figure(figsize=(8, 5))
plt.hist(df["approx_cost"], bins=20)
plt.xlabel("Approximate Cost for Two People")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurant Cost")
plt.tight_layout()
plt.show()
```

```python
# Pie chart for online-order availability.
plt.figure(figsize=(6, 6))
plt.pie(online_counts, labels=online_counts.index, autopct="%1.1f%%", startangle=90, colors=["#1f77b4", "#ffcc99"])
plt.title("Online-Order Availability Distribution")
plt.show()
```
